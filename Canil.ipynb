```python
from dotenv import load_dotenv
import os

# Caminho correto do seu arquivo .env
load_dotenv("/content/drive/MyDrive/Canil/.env")

SUPABASE_URL = os.getenv("SUPABASE_URL")
SUPABASE_KEY = os.getenv("SUPABASE_KEY")

print("URL:", SUPABASE_URL)
print("KEY (início):", SUPABASE_KEY[:10] + "...")
```

```python
!pip install supabase

from supabase import create_client, Client

supabase: Client = create_client(SUPABASE_URL, SUPABASE_KEY)

# Exemplo: buscar todos os registros de uma tabela
resposta = supabase.table("animais").select("*").execute()
print(resposta.data)
```

```python
novo_animal = {
    "Nome": "Rex",
    "Especie": "Cachorro",
    "Raca": "Labrador",        # Campo opcional
    "Porte": "Médio",          # Campo opcional
    "Sexo": "M",               # M para macho, F para fêmea
    "Idade": 4,
    "Cor": "Caramelo",          # Campo opcional
    "DataEntrada": "2025-04-29", # Data personalizada ou pode ser omitido para usar a data atual
    "Status": "Disponível"     # Campo opcional
}

# Inserção no banco de dados
resposta = supabase.table("Animais").insert(novo_animal).execute()

# Resultado da inserção
print(resposta.data)
```
